```python
# encoding 
# mapping
df['burnout_risk']=df['burnout_risk'].map({'Low':0,'Medium':1,'High':2})
```

```python
df['burnout_risk'].unique()
```

```python
#encoding
df=pd.get_dummies(df,columns=['day_type'],dtype='int')
df.columns
```

```python
df.dtypes
```

```python
# label the data
x=df.drop('burnout_risk',axis=1)
y=df['burnout_risk']
```

```python
y.head()
```

```python
# splitting the data for the training and testing
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42) 
```

```python
print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
from sklearn.preprocessing import StandardScaler,MinMaxScaler
sc=StandardScaler()
x_train=sc.fit_transform(x_train) 
x_test=sc.transform(x_test)   
```

```python
x_train
```

```python
x_test
```

```python
df['burnout_risk'].value_counts()
```

```python
from imblearn.over_sampling import SMOTE
smt=SMOTE()
x_train,y_train=smt.fit_resample(x_train,y_train)
```

```python
y_train.value_counts()
```

```python
#model building
#logistic algorithm
from sklearn.linear_model import LogisticRegression
lr=LogisticRegression()
lr.fit(x_train,y_train)
```

```python
lr_pred=lr.predict(x_test)
lr_pred
```

```python
#model training
#for accuracy---to check accuracy of model
from sklearn.metrics import accuracy_score,confusion_matrix
lr_acc=accuracy_score(y_test,lr_pred)
print(lr_acc)
```

```python
import seaborn as sns
```

```python
cm=confusion_matrix(y_test,lr_pred)
print(cm)
plt.figure(figsize=(8,6))
sns.heatmap(cm,annot=True,fmt='d',cmap='Blues',cbar=False,xticklabels=['Predicted Class 0','Predicted Class 1','Predicted Class 2'],yticklabels=['Actual Class 0',
                                'Actual Class 1','Actual Class 2'])
plt.xlabel("Predicted Labels")
plt.ylabel("Actual Labels")
plt.title("Confusion Matrix for Logistic")
plt.show()
```

```python
#decision tree algorithm
from sklearn.tree import DecisionTreeClassifier
dtc=DecisionTreeClassifier()
dtc.fit(x_train,y_train)
```

```python
dtc_pred=dtc.predict(x_test)
dtc_pred
```

```python
#model training
#for accuracy---to check accuracy of model
from sklearn.metrics import accuracy_score,confusion_matrix
dtc_acc=accuracy_score(y_test,dtc_pred)
print(dtc_acc)
```

```python
cm=confusion_matrix(y_test,dtc_pred)
print(cm)
plt.figure(figsize=(8,6))
sns.heatmap(cm,annot=True,fmt='d',cmap='Blues',cbar=False,xticklabels=['Predicted Class 0','Predicted Class 1','Predicted Class 2'],yticklabels=['Actual Class 0',
                                'Actual Class 1','Actual Class 2'])
plt.xlabel("Predicted Labels")
plt.ylabel("Actual Labels")
plt.title("Confusion Matrix for Logistic")
plt.show()
```

```python
#randomforest algorithm
from sklearn.ensemble import RandomForestClassifier
rfc=RandomForestClassifier()
rfc.fit(x_train,y_train)
```

```python
rfc_pred=rfc.predict(x_test)
rfc_pred
```

```python
with open("dtc.pkl","rb") as f:
    model=pickle.load(f)
    print(model)
```

```python
def predict(values):
    values=scaler.transform([values])
    result=model.predict(values)
    if result==[0]:
        print("Low risk")
    elif result==[1]:
        print("Medium risk")
    else:
        print("High risk")
```
